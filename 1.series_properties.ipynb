# What is a Pandas Series?

A Pandas Series is a simple and powerful structure for storing
one-dimensional labeled data in Python.

- A Series is a one-dimensional labeled list of data.
- It stores values together with an index, so each value has a position label.
- Multiple Series can be combined later to build a DataFrame.

```python
import numpy as np
import pandas as pd
print(np.__version__)
print(pd.__version__)
```

```python
salary_values = np.array([25000,32000,0,41000,38000,29000,0,50000,47000,36000])
salary_series = pd.Series(salary_values, name = "Salary")
salary_series
```
