```python
#Slicing on 2D array
import numpy as np
city_array=np.array(["Mumbai","Delhi","Banglore","Hyderabad","Chennai",
                     "Pune","Jaipur","Ahmedbad","Kolkata","Lucknow"

])
city_array=city_array.reshape(2,5)
```

```python
# Reshaping the array into 2 rows and 5 columns
# Original array has 10 cities

#          Column
#          0        1          2          3          4
#       +--------+--------+----------+----------+--------+
# Row 0 | Mumbai | Delhi  | Bengaluru | Hyderabad| Chennai |
#       +--------+--------+----------+----------+--------+
# Row 1 | Pune   | Jaipur | Ahmedabad | Kolkata  | Lucknow |
#       +--------+--------+----------+----------+--------+

# Positive indexing:
# Row 0 -> city_array[0]
# Row 1 -> city_array[1]
#
# Example:
# city_array[0, 0] -> 'Mumbai'
# city_array[1, 4] -> 'Lucknow'

# Negative indexing:
# Row -2 -> first row
# Row -1 -> last row
#
# Example:
# city_array[-1, -1] -> 'Lucknow'
# city_array[-2, -5] -> 'Mumbai'
```

```python
print(city_array[1,3])
print(city_array[1,-1])
print(city_array[:,1:4])
print(city_array[1,1:])
print(city_array[1,:])
print(city_array[:,-1])
print(city_array[:,4])
print(city_array[1,4::-1])#for a row in reverse order
print(city_array[1::-1,3])#for a column in reverse order
```
