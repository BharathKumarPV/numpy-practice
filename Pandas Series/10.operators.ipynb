```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series(salary, index=department, name="Salary")
salary_series
```

```python
#using python comparision operator
print(salary_series == 0)
print(salary_series > 50000)
print(salary_series >= 50000)
print(salary_series < 50000)
print(salary_series <= 48000)
```

```python
#using pandas opearotors
print(salary_series.eq(0))
print(salary_series.ne(0))
print(salary_series.gt(50000))
print(salary_series.ge(50000))
print(salary_series.lt(50000))
print(salary_series.le(48000))
```

```python
print(salary_series.loc[salary_series == 0])
print(salary_series.loc[salary_series > 50000])
print(salary_series.loc[salary_series >= 50000])
print(salary_series.loc[salary_series < 50000])
print(salary_series.loc[salary_series <= 48000])
```

```python
#Filtering by department
m = salary_series.index.isin(["developer", "sales"])
print(salary_series.loc[m])
```
