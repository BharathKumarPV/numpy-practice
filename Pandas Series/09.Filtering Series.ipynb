```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series(salary, index=department, name="Salary")
salary_series
```

```python
salary_series > 50000
```

```python
salary_series.loc[salary_series > 50000]
```

```python
mask = (salary_series >= 45000) & (salary_series < 60000)
salary_series.loc[mask]
```

```python
print(salary_series.loc[salary_series==0])
print(salary_series.loc[salary_series!=0])
print(salary_series.loc[salary_series<=48000])
print("----------")
print(salary_series[salary_series>50000])
```

```python
print(salary_series.loc[(salary_series<45000) | (salary_series > 50000)])
print(salary_series.loc[~(salary_series==0)])
print("----------")
print(salary_series.loc[(salary_series.between(45000, 60000))])
print("------------")
print(salary_series[(salary_series.between(45000, 60000))])
```

```python
print(salary_series.loc[salary_series.index.isin(["developer","sales"])])
```

```python
print(salary_series.loc[salary_series.isin([0, 61000, 45000, 10000])])#no one gets salary as 10000 but it doesnt give error if it sodent exist it return the value which are there
```

```python
mask = salary_series > 50000
print(mask)
print("--------")
print(mask.sum())
print(mask.any())#atleast one true(boolean answer)
print(mask.all())#all should be true(boolean answer)
print(len(salary_series.loc[mask]))
```

```python
salary_series.loc[salary_series<=50000]=50000#with loc
print(salary_series)
print("-----")
salary_series[salary_series<=50000]=50000#without loc
salary_series
```
