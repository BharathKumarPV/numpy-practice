```python
import numpy as np
import pandas as pd
```

```python
salary_values = np.array([25000,32000,41000,38000,29000,50000,47000,36000])
departments = ["developer", "marketing","sales","hr","finance","support","support","operations"]
salary_series=pd.Series(salary_values, index = departments, name="salary")
salary_series
```

```python
print(salary_series.loc["developer"])
print(salary_series.loc["hr"])
print(salary_series.loc["support"])
print(salary_series.loc["marketing":"sales"])
print(salary_series.loc[["developer","sales","support"]])

```

```python
print(salary_series.loc[:"hr"])
print(salary_series.loc["sales":])
print(salary_series.loc["developer":"finance":2])
print(salary_series.loc[salary_series >40000])
print(salary_series.loc[salary_series == 0000])
```
