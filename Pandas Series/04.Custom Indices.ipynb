```python
import numpy as np
import pandas as pd
salary_values = np.array([25000,32000,0,41000,38000,29000,0,50000,47000,36000])
salary_series = pd.Series(salary_values, name = "Salary")
salary_series
```

```python
salary_values = np.array([25000,32000,41000,38000,29000,50000,47000,36000])
departments = ["developer", "marketing","sales","hr","finance","support","support","operations"]
salary_series=pd.Series(salary_values, index = departments, name="salary")
salary_series
```

```python
#Indexing
salary_series['hr']
```

```python
print(salary_series["developer"])
print(salary_series["sales"])
print(salary_series["support"])

```

```python
print(salary_series[["developer","sales"]])
```
