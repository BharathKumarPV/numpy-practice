```python
import numpy as np
import pandas as pd
```

```python
salary_values = np.array([25000,32000,41000,38000,29000,50000,47000,36000])
departments = ["developer", "marketing","sales","hr","finance","support","support","operations"]
salary_series=pd.Series(salary_values, index = departments, name="salary")
salary_series
```

```python
print(salary_series.iloc[0])
print(salary_series.iloc[2])
print(salary_series.iloc[-1])
print(salary_series.iloc[1:4])
print(salary_series.iloc[[0, 3, 5]])
```

```python
print(salary_series.iloc[:3])
print(salary_series.iloc[-3:])
print(salary_series.iloc[::2])
print(salary_series.iloc[::-1])
print(salary_series.iloc[1:5:2])

```

Updating Values

```python
salary_series.iloc[0]=100000 #updating always takes place in original series(in place)
salary_series
```

```python
print(salary_series.iloc[1])
print(salary_series.loc["marketing"])
```

```python
#IndexError: position 10 does not exist, the series has only 6 items
salary_series.iloc[10]
```

```python
#TypeError: iloc works with positions only, it doesn't accept index labels
salary_series.iloc("hr")
```
