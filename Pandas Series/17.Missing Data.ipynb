```python
import pandas as pd
import numpy as np
```

```python
departments = np.array(["HR", "Sales", "Tech", "Finance", "Marketing",
                        "Support", "Design", "Legal", "Admin", "R&D"])

salaries = np.array([55000, np.nan, 72000, 68000, np.nan,
                     55000, 61000, 58000, np.nan, 72000])

salary_series = pd.Series(salaries, index=departments, name="Salary")
salary_series
```

```python
salary_series + 2000
```

```python
salary_series.isna()
```

```python
#shows which department have a missing salary
salary_series[salary_series.isna()]
```

```python
salary_series.notna()#which is not not null will be true
```

```python
salary_series.isna().sum()
```

```python
salary_series.hasnans#scans full data gives True if data contains null values its not a function its a property
```

```python
salary_series.value_counts()
```

```python
salary_series.value_counts(dropna = False)
```

```python
salary_series.dropna()#remove nan values and (all this are not inplace)
```

```python
salary_series.dropna().reset_index()
```

```python
salary_series.fillna(0)#fill nan values with 0
```

```python
salary_series.fillna(round(salary_series.mean(),2))
```

```python
salary_series.fillna(salary_series.median())
```
