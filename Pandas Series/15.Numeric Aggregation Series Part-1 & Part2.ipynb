```python
import pandas as pd
import numpy as np

department = ["engineering", "marketing", "sales", "hr", "finance", "support",
              "design", "operations", "legal", "research", "admin", "security"]

salary_series = pd.Series([52.0, 47.5, 61.0, np.nan, 58.5, 43.0,
                           49.5, 54.0, np.nan, 66.0, 41.5, np.nan],
                          index=department, name="salary")

salary_series
```

```python
salary_series.count()
```

```python
# size counts every row, including the missing ones
salary_series.size
```

```python
salary_series.isna()
```

```python
salary_series.isna().sum()#to count sum od empty values
```

```python
salary_series.sum()
```

```python
salary_series.mean()
```

```python
salary_series.median()
```

```python
salary_series.mode()
```

```python
salary_series.min()
```

```python
salary_series.max()
```

```python
#skipna = False stops pandas from ignoring NaN
salary_series.mean(skipna =False)
```

```python
#filling NaN with 0 pulls the average down which is why missing is kept as NaN
salary_series.fillna(0).mean()
```

```python
round(salary_series.mean(), 2)#rounds upto two decimal
```
