```python
import numpy as np
import pandas as pd

salary = np.array([45000, 52000, 0, 61000, 48000])

department = np.array([
    "developer", "marketing",
    "intern", "sales", "support"
])

salary_series = pd.Series(
    salary,
    index=department,
    name="Salary"
)

salary_series
```

Sort doesnt change inplace it creates another new series and original stays Unchanged

```python
salary_series.sort_index()#index order it come in alphabetical order
```

```python
salary_series.sort_values()#check values and sorts according to it
```

```python
print(salary_series.sort_values(ascending=False))
print("-----------")#note the diff it doesnt affect original series
print(salary_series)
```

```python
salary_series.nlargest(2)#returns top 2 based on values
```

```python
salary_series.nsmallest(2)
```

```python
print(salary_series)
print("---------")
salary_series.rank()#gives value which has least will get rank 1
```

```python
print(salary_series)
print("---------")
salary_series.rank(ascending=False)#now this gives values which has highest get rank 1
```

To change in place we were using series_salary = salary_series.sort()
there is another method easiest which changes inplace
see in below code

```python
# # salary_series = salary_series.sort_values()
# # print(salary_series)
# # print('----------')
salary_series.sort_values(inplace=True)
salary_series
```
