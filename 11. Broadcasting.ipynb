# Broadcasting

## What is Broadcasting?

Broadcasting lets NumPy apply a smaller array to a bigger one automatically when their shapes are compatible.

## Rule Used in Broadcasting

A dimension is compatible if:

- The sizes match.
- Or if one of the dimensions is `1`.

## How Broadcasting Works

Consider the original array:

```python
test_array = np.array([
    [1, 2, 3],
    [1, 2, 3],
    [1, 2, 3]
])

````python
test_array = np.array([[1,2,3],
                       [1,2,3],
                       [1,2,3]])
result = test_array * [1,2,3]
result
````

````python
test_array = np.array([[1,2,3],
                       [1,2,3],
                       [1,2,3]])
result = test_array + 1
result
````

````python
test_array = np.array([[1,2,3],
                       [1,2,3],
                       [1,2,3]])
result = test_array + np.array([1,2,3]).reshape(3,1)
result
````

````python
test_array + np.array([2, 1])
````
