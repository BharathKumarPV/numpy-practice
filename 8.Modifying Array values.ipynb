```python
import numpy as np
scores = np.array([45,67,72,59,88,34])
```

```python
#Change One Value
scores[0]=50
scores
```

```python
#Replace a Slice
scores[1:3]=[70,75]
scores
```

```python
#Update with a condition
scores[scores<40]=40
scores
```

```python
#Update Selected positions
scores[[3,4]]=[60,90]
scores
```
