```python
import numpy as np
```

```python
sales = np.array([120,280,150,180,210,190,170])
sales
```

```python
#1. Total sales
sales.sum() #adds all sales value
```

```python
#2. Average sales
sales.mean() #gives the average sale value
```

```python
#3. Highest Sales
sales.max() #finds the highest sale
```

```python
#4. Lowest Sales
sales.min() #finds the min sale
```

```python
#2D Array
sales1 = np.array([[120, 150, 180, 200],
                   [100, 170, 160, 210],
                   [140, 130, 190, 220]])
sales1
```

```python
#1. Whole array total
sales1.sum()
```

what does axis mean?
* axis = 0 -> combine down the rows and get one result for column
* axis = 1 -> combine across the columns and get the one result per row

```python
#2. Column wise sum
sales1.sum(axis=0)
```

```python
#3. Row wise sum
sales1.sum(axis = 1)
```

```python
#4. Column wise average
sales1.mean(axis = 0)
```

```python
#5. Row-wise average
sales1.mean(axis = 1)
```

```python
#6. Highest Sale per store
sales1.max(axis = 1 )
```
