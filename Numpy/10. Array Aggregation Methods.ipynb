```python
import numpy as np
```

```python
sales = np.array([120,280,150,180,210,190,170])
sales
```

```python
#1. Total sales
sales.sum() #adds all sales value
```

```python
#2. Average sales
sales.mean() #gives the average sale value
```

```python
#3. Highest Sales
sales.max() #finds the highest sale
```

```python
#4. Lowest Sales
sales.min() #finds the min sale
```

```python
#2D Array
sales1 = np.array([[120, 150, 180, 200],
                   [100, 170, 160, 210],
                   [140, 130, 190, 220]])
sales1
```

```python
#1. Whole array total
sales1.sum()
```

what does axis mean?
* axis = 0 -> combine down the rows and get one result for column
* axis = 1 -> combine across the columns and get the one result per row

```python
#2. Column wise sum
sales1.sum(axis = 0)
```

```python
#3. Row wise sum
sales1.sum(axis = 1)
```

```python
#4. Column wise average
sales1.mean(axis = 0)
```

```python
#5. Row-wise average
sales1.mean(axis = 1)
```

```python
#6. Highest Sale per store
sales1.max(axis = 1)
```

```python
#7. Lowest sale per store
sales1.min(axis = 1)
```

Median and Percentiles

```python
sales2 = np.array([120, 150, 180, 200, 210, 220, 260])
sales2
```

```python
#1. Median
np.median(sales2)
```

```python
#2. Percentile
np.percentile(sales2, 75)#75% of values are at or below this number
```

```python
np.percentile(sales2, [25,50,75])#50th perecntile is called as median
```

```python
sales = np.array([150, 200, 150, 300, 250, 200, 350, 150])
```

```python
# Unique values: distinct values, sorted
np.unique(sales)
```

```python
#inique values and how many times it appears
np.unique(sales, return_counts=True)
```

```python
#square root of each value (element wise)
np.sqrt(sales)
```

```python
sales = np.array([120, 200, 150, 100, 210, 190, 170])
```

```python
# Standard deviation: how spread out the values are
sales.std()
```

```python
# Position (index) of the highest and lowest sale
print(sales.argmax(), sales.argmin())
```

```python
#Running Total or cummulative : each value = sum of all values up to that point
sales.cumsum()
```

```python
#in place sort
sales.sort()
sales
```

```python
sorted_sales = np.sort(sales)
print(sorted_sales)
print(sales)
```

```python
#Round each value to 2 decimal places
np.round(np.sqrt(sales),2)
```

```python
#Absolute value: remove the minus sign (e.g. profit/loss -> size of change)
changes = np.array([-20, 15, -5, 30, -10])
np.abs(changes)
```
