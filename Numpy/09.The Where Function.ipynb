```python
# THE WHERE FUNCTION

# The np.where() function is a powerful tool in NumPy.(it always gives a new array)
# It evaluates a condition element-by-element and returns one value where the condition is True,
# and another where it is False.

# --------------------------------------------------
# SYNTAX
# --------------------------------------------------

# np.where(condition, value_if_true, value_if_false)

# --------------------------------------------------
# CONDITION
# --------------------------------------------------

# A logical expression (or array of booleans) that is evaluated
# element-wise.
# Can be any comparison like
# >, <, ==, !=, >=, <=, etc.

# --------------------------------------------------
# VALUE_IF_TRUE
# --------------------------------------------------

# The value (or array) returned at positions where the
# condition is True.

# --------------------------------------------------
# VALUE_IF_FALSE
# --------------------------------------------------

# The value (or array) returned at positions where the
# condition is False.

# --------------------------------------------------
# KEY POINTS
# --------------------------------------------------

# ✓ Operates element-wise on arrays (vectorized).
#
# ✓ Both value_if_true and value_if_false can be scalars or arrays
#   (must be broadcastable to the condition's shape).
#
# ✓ A fast and efficient alternative to writing loops or
#   multiple if-else statements.

# --------------------------------------------------
# WHY IT'S USEFUL
# --------------------------------------------------

# It lets you make choices across entire arrays in a single line
# of code — concise, readable, and highly efficient.
```

```python
#1.Convert grades pass if>=50 else fail
import numpy as np
scores = np.array([45,67,32,90,58])
result=np.where(scores>=50,"Pass","Fail")
result
```

```python
#2.Make numbers positive(replace negative with 0)
nums = np.array([-4,-1,3-7,8,0])
positive_nums = np.where(nums < 0 , 0 , nums)
positive_nums
```

```python
#3. Aplly Discount: 10% off if price>100
prices=np.array([80,120,150,90,200])
final_prices=np.where(prices>100,prices*0.9,prices)
final_prices
```

```python
#4.Categorize ages
ages = np.array([8,15,22,35,60,70])
category = np.where(ages<18,"Child","Adult")
category
```

```python
nums = np.array([1,2,3,4,5,6])
labels = np.where(nums % 2 == 0, "Even","Odd")
labels
```

```python
#6. Cap Values at a maximum(Upper bound)
values = np.array([10,25,50,75,120])
capped = np.where(values > 100 , 100 , values)
capped
```

```python
# Q1. Weather report: label each day "Hot" if temperature > 35, else "Normal"
tempartures = np.array([28,36,41,33,30,38,25])
label = np.where(tempartures > 35 , "Hot", "Normal")
label
```

```python
#Q2. Absent students are marked -1: replace -1 with 0 and keep other marks as they are
marks = np.array([78,-1,56,89,-1,45])
final = np.where(marks == -1 , 0 , marks)
final
```

```python
#Q3. Free delivery: delivery charge is  0 if order amount >= 499 ,  else 40
# Then find the total amount to pay (order + delivery charges)
orders = np.array([250,499,799,120,1500])
final_payment = np.where(orders>=499 , orders, orders + 40)
final_payment
```

```python
#Q4. Salary hike: 20% hike if salary < 30000 , else 10% hike. Find the new salaries
salaries = np.array([25000, 450000, 18000, 60000, 30000])
new_salaries = np.where(salaries < 30000 , salaries * 1.2, salaries * 1.10)
new_salaries
```

```python
#   Q5. Phone battery status (use two arrays together)
# . . . . "charging"--->if the phone is pluggged in
# . . . ."Charging Now"-->if NOT plugged in AND battery <20
# . . . ."OK"------------>Otherwise
battery = np.array([15,80,5,45,19,10])
is_charging = np.array([False,False,True,False,False,True])
status = np.where(is_charging == True , "Charging", np.where(battery<20 , "Charge now", "OK"))
status
```

```python
# Q6. (Challenge) Stock report using yesterday's and today's prices:
# i) First find the % change = (today - yesterday) / yesterday * 100
# ii) Then Label1: "Big Jump" if change > 5, "Up" if change > 0,
#     "Down" if change < 0, else "No Change"
# Hint: put one np.where inside another

yesterday = np.array([200, 150, 320, 90, 500, 75])
today     = np.array([215, 150, 310, 93, 500, 72])
change = (today - yesterday) / yesterday * 100
change
```

```python
label1 = np.where(change > 5 , "Big Jump",
                  np.where(change > 0 , "Up",
                  np.where(change < 0 , "Down", "No change")))
label1        
```
