```python
import pandas as pd

students = ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Sana", "Vikram", "Zoya"]

marks = pd.Series([92, 78, 35, 66, 88, 49, 57, 95], index=students, name="marks")

marks
```

```python
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"
```

```python
grades = marks.apply(get_grade)
grades
```

```python
grades.value_counts()
```
